## Core, byte-identical across every notebook

```python
# ============================================================================
# CORE  -  Equations 1 to 13 of the PNAS manuscript (Methods numbering).
# Byte-identical cell in every notebook, verified by hash (roadmap step 6.4).
# ============================================================================
# Eq 1   p_i = c_i / sum_j c_j                       coupling share
# Eq 2   H = - sum_i p_i log2 p_i                     control entropy (avg information)
# Eq 3   p_i = C_i / sum_j C_j,  A C = lambda C       eigenvector weighting
# Eq 4   J = H / log2 N                               evenness
# Eq 5   D_eff = 2^H                                  effective count of DOF
# Eq 6   D_eff / N = 2^(-dH),  dH = log2 N - H        retained fraction / compression
# Eq 7   H <= log2 N                                  the bound (Gibbs / Jensen)
# Eq 8   dH/deps = log2(p_j / p_i)                    transfer gradient, concentration lowers H
# Eq 9   H = H2(phi) + phi log2 h + (1-phi) log2(N-h) leader-follower closed form
# Eq 10  kappa = <k^2> / <k>                          degree heterogeneity
# Eq 11  r                                            degree assortativity
# Eq 12  dh = log2(sigma_A / sigma_B)                 outcome entropy difference
# Eq 13 (Hebbian) lives in s4, since it needs the parent model's activations.
#
# I(p) = log2(1/p) is the summand of Eq 2 (Introduction Eq 2).
# Compression dH = log2 N - H is the bits saved (Introduction Eq 5), used by Eq 6.
# ============================================================================
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

CORE_VERSION = "v5 / 24 July 2026 (Methods Eq 1-13)"

def share(c):
    """Eq 1. Normalise any non-negative vector into coupling shares."""
    c = np.asarray(c, dtype=float).ravel()
    if np.any(c < 0):
        raise ValueError("coupling values must be non-negative")
    t = c.sum()
    if t <= 0:
        raise ValueError("total coupling is zero")
    return c / t

def information(p):
    """Information of a share, in bits, the summand of Eq 2. Small share -> large information."""
    p = np.asarray(p, dtype=float)
    out = np.full(p.shape, np.inf)
    np.log2(1.0 / p, out=out, where=(p > 0))
    return out

def shannon_entropy(c):
    """Eq 2. H as the average of the information over the distribution. Input may be raw."""
    p = share(c)
    p = p[p > 0]
    return float((p * information(p)).sum())

def max_entropy(n):
    """Eq 7 (right side). The ceiling log2 N, attained only by the uniform distribution."""
    if n < 1:
        raise ValueError("n must be at least 1")
    return float(np.log2(n))

def evenness(h, n):
    """Eq 4. J = H / log2 N. Compare systems of different size with this."""
    m = max_entropy(n)
    return 1.0 if m == 0 else float(h / m)

def effective_dof(h):
    """Eq 5. D_eff = 2^H, a number of equally coupled degrees of freedom."""
    return float(2.0 ** h)

def compression(H, N):
    """dH = log2 N - H (Introduction Eq 5). Bits saved against even coupling.
    Regular gives 0 exactly; the scale-free ensemble gives 0.326 at N = 100.
    """
    return float(np.log2(N) - H)

def retained_fraction(H, N):
    """Eq 6. 2^(-dH) = D_eff / N, the fraction of degrees of freedom still to be
    specified. Equals 2^H / N. Scale-free gives 0.798, the 79.8 out of 100.
    """
    return float(2.0 ** (-(np.log2(N) - H)))

def coupling_shares(G, weight="degree"):
    """Eq 1 with c = degree, or Eq 3 with c = eigenvector centrality."""
    nodes = list(G.nodes())
    if weight == "degree":
        v = np.array([G.degree(n) for n in nodes], dtype=float)
    elif weight == "eigenvector":
        c = nx.eigenvector_centrality_numpy(G)
        v = np.abs(np.array([c[n] for n in nodes], dtype=float))
    else:
        raise ValueError("weight must be 'degree' or 'eigenvector'")
    return share(v)

def control_entropy(G, weight="degree"):
    """Eq 2 applied to Eq 1 or Eq 3."""
    return shannon_entropy(coupling_shares(G, weight=weight))

def strength_shares(W):
    """Eq 1 for a weighted structure: c_i = node strength = sum_j W_ij."""
    W = np.asarray(W, dtype=float)
    if W.ndim != 2 or W.shape[0] != W.shape[1]:
        raise ValueError("W must be square")
    return share(W.sum(axis=1))

def transfer_gradient(p, i, j):
    """Eq 8. dH/deps = log2(p_j / p_i) when eps moves coupling FROM j TO i.
    Negative whenever p_i > p_j, so any transfer toward an already-strong
    element strictly lowers H. i is the element that receives coupling.
    """
    p = np.asarray(p, dtype=float)
    return float(np.log2(p[j] / p[i]))

def _binary_entropy(phi):
    """H2(phi) = -phi log2 phi - (1-phi) log2(1-phi), with the 0 log 0 = 0 rule."""
    phi = float(phi)
    if phi <= 0.0 or phi >= 1.0:
        return 0.0
    return float(-phi * np.log2(phi) - (1.0 - phi) * np.log2(1.0 - phi))

def two_level_entropy(N, h, phi):
    """Eq 9. H = H2(phi) + phi log2 h + (1-phi) log2(N-h).
    h leaders share phi equally and N-h followers share (1-phi) equally.
    Returns log2 N exactly when phi = h/N, the built-in unit test.
    """
    if not (1 <= h < N):
        raise ValueError("need 1 <= h < N")
    if not (0.0 <= phi <= 1.0):
        raise ValueError("phi must be in [0, 1]")
    lead = phi * np.log2(h) if h > 0 else 0.0
    return float(_binary_entropy(phi) + lead + (1.0 - phi) * np.log2(N - h))

def degree_heterogeneity(G):
    """Eq 10. kappa = <k^2> / <k>, the parent paper's normalisation.

    NOTE ON NORMALISATION. The handwritten note of 23 July writes the ratio as
    <k^2>/<k>^2. That differs from this by a factor <k>, and the parent paper's
    published values (6.84, 6.09, 9.67) are on THIS definition. Table 1 of the
    manuscript uses this one so the two papers stay comparable. Convert with
    kappa_alt = kappa / <k> if a source uses the other.
    """
    k = np.array([d for _, d in G.degree()], dtype=float)
    if k.mean() <= 0:
        return float("nan")
    return float((k ** 2).mean() / k.mean())

def degree_assortativity(G):
    """Eq 11. Pearson correlation of degrees at the two ends of an edge.

    Undefined for a regular graph, where the degrees carry no variance. Returns
    nan there rather than a fabricated zero.
    """
    k = np.array([d for _, d in G.degree()], dtype=float)
    if k.std() == 0 or G.number_of_edges() == 0:
        return float("nan")
    try:
        r = nx.degree_assortativity_coefficient(G)
    except (ValueError, ZeroDivisionError):
        return float("nan")
    return float(r) if np.isfinite(r) else float("nan")

def gini(x):
    """Gini coefficient, matching the parent paper's definition."""
    x = np.sort(np.asarray(x, dtype=float))
    n = len(x)
    if x.sum() == 0:
        return 0.0
    idx = np.arange(1, n + 1)
    return float((2.0 * (idx * x).sum()) / (n * x.sum()) - (n + 1.0) / n)

GAUSS_CONST = 0.5 * np.log2(2.0 * np.pi * np.e)

def outcome_entropy_gaussian(sigma):
    """h = 0.5 log2(2 pi e sigma^2). A DIFFERENTIAL entropy, not Eq 2.

    It shares the unit and the reading of predictability with control entropy
    and is a different quantity on a different object. Only differences on a
    common scale are interpreted, which is why Eq 12 exists.
    """
    s = np.asarray(sigma, dtype=float)
    if np.any(s <= 0):
        raise ValueError("sigma must be positive")
    return GAUSS_CONST + np.log2(s)

def delta_outcome_entropy(sigma_a, sigma_b):
    """Eq 12. The additive constant and the unit both cancel."""
    return float(np.log2(float(sigma_a) / float(sigma_b)))

def predictability_ratio(dh):
    """R = 2^dh, the predictability ratio that follows Eq 12."""
    return float(2.0 ** float(dh))

print("core loaded:", CORE_VERSION)
```

## The four matched ensembles, and the two quantities from control theory

```python
# =========================================================================
# THE FOUR MATCHED ENSEMBLES, AND THE TWO QUANTITIES FROM CONTROL THEORY
# =========================================================================
# Generators as published (Supplementary Note 1) and as Figure 3 draws them:
# 100 elements, exactly 300 links in every realisation, seeds 42 to 141.
# Driver nodes are counted on the full generated graph, as kappa and the
# assortativity are. Shares and average controllability are taken on the largest
# connected component, where the eigenvector shares of Table 1 are defined.
# Every definition below is the one in s11 (Supplementary Note 6).
N, K_MEAN, N_REAL, SEED0 = 100, 6, 100, 42
M_LINKS = N * K_MEAN // 2                # 300 links in every realisation
BA_M, BA_N0 = 3, 12                      # 36 + 3 x 88 = 300


def make(kind, seed):
    if kind == 'regular':     return nx.watts_strogatz_graph(N, K_MEAN, 0.0, seed=seed)
    if kind == 'small-world': return nx.watts_strogatz_graph(N, K_MEAN, 0.1, seed=seed)
    if kind == 'random':      return nx.gnm_random_graph(N, M_LINKS, seed=seed)
    if kind == 'scalefree':
        g0 = nx.watts_strogatz_graph(BA_N0, K_MEAN, 0.0, seed=seed)
        return nx.barabasi_albert_graph(N, BA_M, seed=seed, initial_graph=g0)
    raise ValueError(kind)


def giant(G):
    return G if nx.is_connected(G) else G.subgraph(
        max(nx.connected_components(G), key=len)).copy()


KINDS = ['regular', 'small-world', 'random', 'scalefree']
LABEL = {'regular': 'Regular lattice', 'small-world': 'Small-world',
         'random': 'Random', 'scalefree': 'Scale-free'}
GEN = {'regular': r'ring,  $\beta$ = 0', 'small-world': r'rewired,  $\beta$ = 0.1',
       'random': 'Erdős–Rényi,  fixed link count', 'scalefree': 'preferential attachment'}


def driver_nodes_matching(G):
    """Liu et al. (2011). N_D = max(N - |M*|, 1), with M* a maximum matching of
    the directed network, each undirected link taken in both directions."""
    out_side = [('out', n) for n in G.nodes()]
    B = nx.Graph()
    B.add_nodes_from(out_side, bipartite=0)
    B.add_nodes_from([('in', n) for n in G.nodes()], bipartite=1)
    for u, v in G.edges():
        B.add_edge(('out', u), ('in', v)); B.add_edge(('out', v), ('in', u))
    M = bipartite.hopcroft_karp_matching(B, top_nodes=out_side)
    return max(G.number_of_nodes() - sum(1 for key in M if key[0] == 'out'), 1)


def driver_nodes_exact(G, tol=1e-6):
    """Yuan et al. (2013). For a symmetric adjacency matrix the exact count is the
    largest multiplicity of any eigenvalue, confirmed by the rank of lambda I - A."""
    A = nx.to_numpy_array(G, nodelist=sorted(G.nodes()))
    ev = np.sort(np.linalg.eigvalsh(A))
    groups = np.split(ev, np.where(np.diff(ev) > tol)[0] + 1)
    top = max(groups, key=len)
    lam = float(top.mean())
    by_rank = A.shape[0] - np.linalg.matrix_rank(lam * np.eye(A.shape[0]) - A, tol=tol)
    assert by_rank == len(top), (by_rank, len(top))
    return int(len(top)), lam


def average_controllability(G, c=1.0):
    """Gu et al. (2015): the trace of the infinite-horizon controllability Gramian
    with input at one element, [(I - A_hat^2)^-1]_ii for a symmetric matrix, with
    A_hat = A / (c + its largest singular value). In G.nodes() order."""
    nodes = list(G.nodes())
    A = nx.to_numpy_array(G, nodelist=nodes)
    As = A / (c + np.linalg.norm(A, 2))
    return np.diag(np.linalg.inv(np.eye(len(nodes)) - As @ As))


ens = {}
for kind in KINDS:
    H, ndm, nde, r = [], [], [], []
    for s in range(SEED0, SEED0 + N_REAL):
        Gf = make(kind, s); G = giant(Gf)
        p = coupling_shares(G, 'eigenvector')
        H.append(shannon_entropy(p))
        ndm.append(driver_nodes_matching(Gf)); nde.append(driver_nodes_exact(Gf)[0])
        r.append(float('nan') if np.ptp(p) < 1e-12 else
                 float(np.corrcoef(p, average_controllability(G))[0, 1]))
    ens[kind] = {'H': np.array(H), 'match': np.array(ndm), 'exact': np.array(nde),
                 'r': np.array(r)}

# Table 1 comes back to its printed precision, so these are the same 400 networks.
TABLE1 = {'regular': (6.6439, 100.0), 'small-world': (6.5832, 95.9),
          'random': (6.4820, 89.4), 'scalefree': (6.3176, 79.8)}
for kind in KINDS:
    Hm = ens[kind]['H'].mean()
    assert (round(Hm, 4), round(effective_dof(Hm), 1)) == TABLE1[kind], kind

# The archived s11 output, if it sits beside this notebook or in data/, must agree
# row for row.
import os, csv
S11 = next((f for f in ('s11_controllability_checks.csv', 'data/s11_controllability_checks.csv')
            if os.path.exists(f)), None)
if S11:
    rows = list(csv.DictReader(open(S11)))
    for kind in KINDS:
        sub = [x for x in rows if x['topology'] == kind]
        assert [int(x['driver_nodes_matching']) for x in sub] == list(ens[kind]['match'])
        assert [int(x['driver_nodes_exact']) for x in sub] == list(ens[kind]['exact'])
        ref = np.array([float(x['r_share_avgctrl']) for x in sub])
        ok = np.isfinite(ref)
        assert np.allclose(ref[ok], ens[kind]['r'][ok], atol=1e-6)
    print('agrees row for row with', S11)
else:
    print('s11_controllability_checks.csv not found beside the notebook; values recomputed only')
```

## The drawn structures, one minimal set of driver nodes in each, and the two structures of E and G

```python
# =========================================================================
# THE DRAWN STRUCTURES, AND ONE MINIMAL SET OF DRIVER NODES IN EACH
# =========================================================================
# Each column draws the first realisation of its ensemble, seed 42, which is the
# structure Figure 3 draws at 100 elements, on Figure 3's own spring rule, so the
# two figures show the same networks in the same layout. Panel F reads the same
# four realisations.
SHOW = {k: giant(make(k, SEED0)) for k in KINDS}


def drawn_shares(G):
    """The shares the picture is built from, rounded once at the source, as in
    Figure 3, so the same script always renders the same image."""
    return np.round(coupling_shares(G, 'eigenvector'), 12)


SHOW_W = {k: drawn_shares(SHOW[k]) for k in KINDS}


def spread(G, w):
    """Figure 3's spring rule, unchanged: the leading element seeded at the centre
    and the rest on a golden-angle spiral by rank in share, then the spring run
    from there, ties broken on rounded shares, then degree, then label."""
    idx = {n: i for i, n in enumerate(G.nodes())}
    order = sorted(G.nodes(), key=lambda z: (-round(float(w[idx[z]]), 9), -G.degree(z), z))
    n = len(order)
    seed_pos = {}
    for t, node in enumerate(order):
        r = np.sqrt(t / max(n - 1, 1))
        a = t * np.pi * (3.0 - np.sqrt(5.0))
        seed_pos[node] = np.array([r * np.cos(a), r * np.sin(a)])
    pos = nx.spring_layout(G, pos=seed_pos, seed=3, iterations=900, k=2.8 / np.sqrt(n))
    r = np.array([np.hypot(*pos[z]) for z in G.nodes()])
    s = np.sqrt(np.sqrt((r ** 2).mean()) * r.max()) or 1.0
    return {z: (v[0] / s, v[1] / s) for z, v in pos.items()}


POS = {k: spread(SHOW[k], SHOW_W[k]) for k in KINDS}


def driver_set_exact(G, w):
    """One minimal set of driver nodes for the exact count, checked by the
    Popov-Belevitch-Hautus test. For a symmetric A the system is controllable from
    a set D exactly when, for every eigenvalue, the rows of its eigenvector basis
    at D span that eigenspace. Elements are added one at a time, each time the one
    that closes the most of the remaining gap, ties going first to the element
    farthest from those already chosen and then to the lowest share. The set is
    not unique, and the figure draws one valid set."""
    nodes = list(G.nodes()); ix = {n: i for i, n in enumerate(nodes)}
    A = nx.to_numpy_array(G, nodelist=nodes)
    lam, V = np.linalg.eigh(A)
    groups = np.split(np.arange(len(lam)), np.where(np.diff(lam) > 1e-6)[0] + 1)
    bases = [V[:, g] for g in groups]
    need = max(len(g) for g in groups)

    def gap(D):
        return sum(B.shape[1] - (np.linalg.matrix_rank(B[[ix[d] for d in D], :], tol=1e-8) if D else 0)
                   for B in bases)

    order = sorted(nodes, key=lambda z: (round(float(w[ix[z]]), 9), G.degree(z), z))
    D = []
    while gap(D) > 0:
        g0 = gap(D)
        dist = (lambda z: min(nx.shortest_path_length(G, z, d) for d in D)) if D else (lambda z: 0)
        best = min(order, key=lambda z: (gap(D + [z]) - g0 if z not in D else 1, -dist(z), order.index(z)))
        D.append(best)
    assert len(D) == need, (len(D), need)                  # minimal, as Yuan et al. prove
    assert all(np.linalg.matrix_rank(B[[ix[d] for d in D], :], tol=1e-8) == B.shape[1] for B in bases)
    return D


DRIVERS = {k: driver_set_exact(SHOW[k], SHOW_W[k]) for k in KINDS}


def turned(pos, drivers):
    """The same layout turned about its centre so that the input arrows run as
    nearly sideways as they can, a lone driver arriving from the left. A layout
    carries no information in its orientation, so nothing is changed but the
    room the arrows leave for the titles above and the readings below."""
    xy = np.array(list(pos.values())); c = xy.mean(axis=0)
    th = [np.arctan2(*(np.array(pos[d]) - c)[::-1]) for d in drivers]
    best = max(np.deg2rad(np.arange(0, 360)),
               key=lambda f: (round(min(abs(np.cos(t + f)) for t in th), 6), -np.cos(th[0] + f)))
    R = np.array([[np.cos(best), -np.sin(best)], [np.sin(best), np.cos(best)]])
    return {n: tuple(c + R @ (np.array(v) - c)) for n, v in pos.items()}


POS = {k: turned(POS[k], DRIVERS[k]) for k in KINDS}
for k in KINDS:   # the drawn sets carry the exact count of the drawn realisation
    assert len(DRIVERS[k]) == ens[k]['exact'][0] == driver_nodes_exact(make(k, SEED0))[0]

# The scale-free realisation of seed 42, whose elements F draws one by one.
G_SF = SHOW['scalefree']
P_SF = coupling_shares(G_SF, 'eigenvector')
AC_SF = average_controllability(G_SF)
R_SF = float(np.corrcoef(P_SF, AC_SF)[0, 1])
assert abs(R_SF - ens['scalefree']['r'][0]) < 1e-9
SLOPE_SF, ICPT_SF = np.polyfit(P_SF, AC_SF, 1)
K_SF = np.array([G_SF.degree(n) for n in G_SF.nodes()], dtype=float)

# The lower row, F: influence against share for every element of the four
# structures drawn above, with the least-squares line of each uneven one.
INFL = {}
for k in KINDS:
    Gk = SHOW[k]
    pk = coupling_shares(Gk, 'eigenvector'); ack = average_controllability(Gk)
    INFL[k] = {'p': pk, 'ac': ack,
               'fit': None if np.ptp(pk) < 1e-12 else np.polyfit(pk, ack, 1),
               'r': float('nan') if np.ptp(pk) < 1e-12 else float(np.corrcoef(pk, ack)[0, 1])}
assert np.allclose(INFL['scalefree']['p'], P_SF) and np.allclose(INFL['scalefree']['ac'], AC_SF)
assert np.ptp(INFL['regular']['ac']) < 1e-9          # the lattice: one share, one influence

# The lower row, E and G: one coordinative structure drawn at two stages of
# practice, for illustration and not measured. Sixteen neural populations in the
# brain and eight muscles of the arm, the anterior, middle and posterior deltoid,
# the long and short heads of the biceps, the triceps, the brachioradialis and
# the forearm flexors, joined by forty-eight links at each stage, as the four
# ensembles above are matched in size and links. At the initial-like stage every
# element holds four links: each population is coupled to three nearby
# populations and drives one muscle, and each muscle is driven by two
# populations and coupled to the two muscles beside it in a ring. At the
# skilled-like stage the same number of links was laid down by preferential
# attachment, each new link the likelier to join an element the more links it
# already held, so hubs form in the brain and at the long head of the biceps,
# which crosses both shoulder and elbow. The edge lists are written out so the
# drawing does not depend on a solver or a random stream.
NEURAL = [f'n{k}' for k in range(16)]
MUSCLE = ['AD', 'MD', 'PD', 'BL', 'BS', 'TR', 'BR', 'FL']
NEURAL_REL = {'n1': (0.06, 0.06), 'n0': (-0.30, -0.08), 'n2': (0.40, 0.30), 'n3': (-0.08, 0.46),
              'n4': (-0.50, 0.26), 'n5': (0.36, -0.28), 'n6': (-0.06, -0.42), 'n7': (0.68, 0.02),
              'n8': (0.68, 0.44), 'n9': (0.24, 0.70), 'n10': (-0.36, 0.66), 'n11': (-0.74, 0.34),
              'n12': (-0.82, -0.14), 'n13': (-0.50, -0.40), 'n14': (0.16, -0.54), 'n15': (0.58, -0.36)}
INITIAL_LIKE, SKILLED_LIKE = nx.Graph(), nx.Graph()
INITIAL_LIKE.add_nodes_from(NEURAL + MUSCLE); SKILLED_LIKE.add_nodes_from(NEURAL + MUSCLE)
# a 3-regular lattice of short links among the populations, their single
# descending links by the shortest assignment, and the muscles in a ring
INITIAL_LIKE.add_edges_from([
    ('n0', 'n1'), ('n0', 'n13'), ('n0', 'n6'), ('n1', 'n3'), ('n1', 'n5'), ('n10', 'n11'),
    ('n11', 'n12'), ('n12', 'n13'), ('n14', 'n15'), ('n2', 'n7'), ('n2', 'n8'), ('n2', 'n9'),
    ('n3', 'n10'), ('n3', 'n9'), ('n4', 'n10'), ('n4', 'n11'), ('n4', 'n12'), ('n5', 'n14'),
    ('n5', 'n15'), ('n6', 'n13'), ('n6', 'n14'), ('n7', 'n15'), ('n7', 'n8'), ('n8', 'n9'),
    ('n0', 'MD'), ('n1', 'BL'), ('n10', 'MD'), ('n11', 'PD'), ('n12', 'TR'), ('n13', 'PD'),
    ('n14', 'BL'), ('n15', 'BR'), ('n2', 'BR'), ('n3', 'AD'), ('n4', 'TR'), ('n5', 'BS'),
    ('n6', 'AD'), ('n7', 'FL'), ('n8', 'FL'), ('n9', 'BS')] +
    [(MUSCLE[i], MUSCLE[(i + 1) % 8]) for i in range(8)])
# grown by preferential attachment from the triangle of n0, n1 and the long head
# of the biceps
SKILLED_LIKE.add_edges_from([
    ('AD', 'BL'), ('AD', 'n1'), ('AD', 'n11'), ('BL', 'BS'), ('BL', 'FL'), ('BL', 'n0'), ('BL', 'n1'),
    ('BL', 'n12'), ('BL', 'n15'), ('BL', 'n2'), ('BL', 'n7'), ('BR', 'n6'), ('BR', 'n7'), ('BS', 'FL'),
    ('BS', 'n0'), ('MD', 'n0'), ('MD', 'n1'), ('MD', 'n11'), ('MD', 'n2'), ('MD', 'n9'), ('PD', 'TR'),
    ('PD', 'n0'), ('TR', 'n0'), ('TR', 'n1'), ('TR', 'n6'), ('n0', 'n1'), ('n0', 'n14'), ('n0', 'n3'),
    ('n0', 'n4'), ('n0', 'n5'), ('n1', 'n10'), ('n1', 'n12'), ('n1', 'n13'), ('n1', 'n3'), ('n1', 'n6'),
    ('n1', 'n7'), ('n1', 'n8'), ('n1', 'n9'), ('n10', 'n13'), ('n12', 'n14'), ('n12', 'n5'),
    ('n12', 'n8'), ('n13', 'n7'), ('n13', 'n8'), ('n15', 'n2'), ('n15', 'n7'), ('n3', 'n7'),
    ('n4', 'n6')])
# The labels of a structure grown by preferential attachment carry no meaning,
# so two are exchanged to place the element that can drive it alone at the back
# of the brain, as in the initial-like structure, so that in E and in its mirror
# image G the input arrives from the outer edge of the sheet.
SKILLED_LIKE = nx.relabel_nodes(SKILLED_LIKE, {'n12': 'n14', 'n14': 'n12'})
assert INITIAL_LIKE.number_of_edges() == SKILLED_LIKE.number_of_edges() == 48
assert set(dict(INITIAL_LIKE.degree()).values()) == {4}
assert nx.is_connected(INITIAL_LIKE) and nx.is_connected(SKILLED_LIKE)


def single_driver(G, prefer):
    """For a structure whose eigenvalues are all simple, one element drives it
    exactly when every eigenvector is non-zero there (Popov-Belevitch-Hautus).
    Returns the first such element in the order given."""
    nodes = list(G.nodes()); A = nx.to_numpy_array(G, nodelist=nodes)
    lam, V = np.linalg.eigh(A)
    assert np.all(np.diff(lam) > 1e-6), 'a repeated eigenvalue: more than one driver node'
    ok = [n for i, n in enumerate(nodes) if np.all(np.abs(V[i, :]) > 1e-6)]
    return next(n for n in prefer if n in ok)


SCHEME = {}
for name, G in (('initial', INITIAL_LIKE), ('skilled', SKILLED_LIKE)):
    nd_, _ = driver_nodes_exact(G)
    S_ = {'G': G, 'p': coupling_shares(G, 'eigenvector'), 'nd': nd_}
    S_['H'] = shannon_entropy(S_['p']); S_['eff'] = effective_dof(S_['H'])
    # the driver drawn is the valid population nearest the back of the brain
    S_['driver'] = single_driver(G, sorted(NEURAL, key=lambda n: NEURAL_REL[n][0]))
    SCHEME[name] = S_
assert SCHEME['initial']['nd'] == SCHEME['skilled']['nd'] == 1
```

```python
# =========================================================================
# FIGURE
# =========================================================================
# Two rows. The upper row keeps the four columns of Figure 3, A to D, with the
# letters and count axes of C and D at the outer edge, as in Figures 3 and 4.
# Each column reads downward through the
# structure of seed 42, the one Figure 3 draws at 100 elements, with one minimal
# set of driver nodes, each ringed and fed by an input of its own drawn as in Liu
# et al. (2011), and the distribution of the driver-node count over the hundred
# realisations of the ensemble, with the two counts of the ensemble in its upper
# corner. The lower row reads the same coupling as influence. F sets average
# controllability against coupling share for every element of the structures
# above, and E and G draw one small coordinative structure of neural populations
# and arm muscles at two stages of practice, its coupling spread evenly at the
# initial-like stage and laid down by preferential attachment at the skilled-like
# stage. Beneath each structure the hand reaches five times for a target, and a
# double arrow marks the absolute error of the endpoints. G is drawn as the
# mirror image of E, so the two stages face each other across F, and its letter
# sits at the outer edge, as in C and D.
# Elements read log2(p_i N) on the scale Figure 3 prints, so, as in Figure 4,
# the figure carries no colour bar of its own. Every text is black, and the
# panel letters are the only bold type. Every position is in inches from the
# lower left corner of the sheet.
from matplotlib.lines import Line2D
from matplotlib.patches import Ellipse
from matplotlib.legend_handler import HandlerBase
from matplotlib.ticker import NullLocator, FixedLocator

FIG_W_IN = 6.93
MIRROR = (False, False, True, True)

# Columns. The outer edges keep the sheet margins, and, as in the lower row of
# Figure 4, the two inner gaps that carry count labels are wider than the middle.
MARGIN, G_IN, G_MID = 0.385, 0.240, 0.130
COL_W = (FIG_W_IN - 2 * MARGIN - 2 * G_IN - G_MID) / 4
LX = [MARGIN, MARGIN + COL_W + G_IN, MARGIN + 2 * COL_W + G_IN + G_MID,
      MARGIN + 3 * COL_W + 2 * G_IN + G_MID]
CX = [x + COL_W / 2 for x in LX]

# Rows, bottom up. The lower row first, then the upper row lifted above it. The
# structures sit low in their square, since none reaches its lower edge, so the
# title of each distribution runs under them.
BR_Y, BR_H = 0.330, 1.000                  # the plot in F
TB_Y = BR_Y + BR_H + 0.140                 # letters and titles of the lower row
ST_Y, ST_H = TB_Y + 0.470, 0.480           # the distributions, a short strip under each structure
NET_Y = ST_Y + ST_H - 0.060                # the structures
NET_H = 1.240
T_Y = NET_Y + NET_H + 0.030                # letters and titles, generator beneath
FIG_H_IN = T_Y + 0.150

fig = plt.figure(figsize=(FIG_W_IN, FIG_H_IN))
fig.patch.set_facecolor('white')


def box(x, y, w, h):
    return fig.add_axes([x / FIG_W_IN, y / FIG_H_IN, w / FIG_W_IN, h / FIG_H_IN])


def ftext(x, y, s, **kw):
    kw.setdefault('color', TXT)
    return fig.text(x / FIG_W_IN, y / FIG_H_IN, s, **kw)


def heading(letter, x_letter, mirror, x_title, title, sub, y):
    """A panel letter at the outer edge, a title centred and a line beneath it."""
    ftext(x_letter, y, letter, fontsize=9.4, fontweight='bold',
          ha='right' if mirror else 'left', va='bottom')
    ftext(x_title, y, title, fontsize=7.9, ha='center', va='bottom')
    if sub:
        ftext(x_title, y - 0.020, sub, fontsize=6.2, ha='center', va='top')


# ---------------- A to D, the structures, with their driver nodes and inputs ---
LIM = 1.04 * max(np.hypot(*POS[k][n]) for k in KINDS for n in SHOW[k].nodes())
_KMAX = float(max(max(dict(SHOW[k].degree()).values()) for k in KINDS))


def nsize(G):
    """Node area by the links an element holds, one scale across the four, so an
    element of six links is drawn at one size wherever it appears."""
    d = np.array([G.degree(n) for n in G.nodes()], dtype=float)
    return 1.6 + 46.0 * (d / _KMAX)


def ringdist(u, v):
    d = abs(u - v)
    return min(d, N - d)


def zigzag(a, b, n=3, amp=0.0):
    """The zigzag of an input signal from a to b, as Liu et al. draw it: n full
    teeth of half-width amp, centred on the line from a to b."""
    d = b - a; L = np.hypot(*d); v = d / L; nrm = np.array([-v[1], v[0]])
    t = np.linspace(0.0, 1.0, 2 * n + 1)
    off = np.zeros_like(t); off[1:-1:2], off[2:-1:2] = amp, -amp
    return np.array([a + v * L * s + nrm * o for s, o in zip(t, off)])


# Distances along the outward line from a driver node, in inches on the page. A
# numbered label is wider than a single u, so it sits R_NUM further out and
# clears its input by about as much as a single u does.
R_TIP, R_ZIG, R_NODE, R_LAB, R_NUM = 0.058, 0.120, 0.226, 0.292, 0.018
for i, kind in enumerate(KINDS):
    ax = box(CX[i] - NET_H / 2, NET_Y, NET_H, NET_H)
    ax.set_xlim(-LIM * 1.06, LIM * 1.06); ax.set_ylim(-LIM * 1.06, LIM * 1.06)
    ax.set_aspect('equal'); ax.axis('off')
    G, w, pos = SHOW[kind], SHOW_W[kind], POS[kind]
    if kind in ('regular', 'small-world'):
        # the rewired links of the small-world stand out from the ring, as in Figure 3
        near = [(pos[u], pos[v]) for u, v in G.edges() if ringdist(u, v) <= K_MEAN // 2]
        far = [(pos[u], pos[v]) for u, v in G.edges() if ringdist(u, v) > K_MEAN // 2]
    else:
        near, far = [(pos[u], pos[v]) for u, v in G.edges()], []
    ax.add_collection(LineCollection(near, colors=LINK, linewidths=0.30, alpha=0.75, zorder=1))
    if far:
        ax.add_collection(LineCollection(far, colors=NEUTRAL, linewidths=0.45, alpha=0.60, zorder=2))
    nodes = list(G.nodes())
    xy = np.array([pos[n] for n in nodes]); sz = nsize(G)
    ax.scatter(xy[:, 0], xy[:, 1], s=sz, c=node_colour(w, N), edgecolors=NEUTRAL,
               linewidths=0.25, zorder=3)
    # one minimal set of driver nodes, each ringed and fed by an input of its own
    # from outside the structure, along the line from the centre of the layout
    c = xy.mean(axis=0)
    upi = 2 * LIM * 1.06 / NET_H              # layout units per inch on the page
    for j, d in enumerate(DRIVERS[kind]):
        p = np.array(pos[d]); k_ = nodes.index(d)
        ax.scatter([p[0]], [p[1]], s=sz[k_] + 30, facecolors='none', edgecolors=INK,
                   linewidths=0.85, zorder=4)
        u = p - c; u = u / (np.hypot(*u) or 1.0)
        tip, z_end = p + u * R_TIP * upi, p + u * R_ZIG * upi
        z_start, q = p + u * (R_NODE - 0.030) * upi, p + u * R_NODE * upi
        zz = zigzag(z_start, z_end, n=3, amp=0.0135 * upi)
        ax.plot(zz[:, 0], zz[:, 1], color=INK, lw=0.65, solid_joinstyle='miter',
                zorder=5, clip_on=False)
        ax.add_patch(FancyArrowPatch(tuple(z_end), tuple(tip), arrowstyle='-|>', mutation_scale=5.0,
                                     color=INK, lw=0.65, shrinkA=0, shrinkB=0, zorder=5,
                                     clip_on=False))
        ax.scatter([q[0]], [q[1]], s=11.0, c=INK, edgecolors=INK, linewidths=0.3, zorder=6,
                   clip_on=False)
        numbered = len(DRIVERS[kind]) > 1
        lab = f'$u_{{{j + 1}}}$' if numbered else '$u$'
        ax.text(*(p + u * (R_LAB + (R_NUM if numbered else 0.0)) * upi), lab, fontsize=6.4,
                ha='center', va='center', zorder=6, clip_on=False)

# ---------------- A to D, the distribution over the hundred realisations ---------
# The frame of the lower row of Figure 4 opened at the top, so each distribution
# runs straight on from the structure above it: a line below only, dotted rules,
# the counts on the outer side of each half, left in A and B and right in C and
# D, and three labelled ticks on each axis. Each count is a stem with a head, the
# count by maximum matching to the left of each value in the grey Figure 4 gives
# a reference, and the exact count to the right with its head in colour, warmer
# with more driver nodes, from yellow at one to dark red at ten. The two counts
# of the ensemble, the effective count 2^H and the mean number of driver nodes by
# each count, are written in the empty upper corner.
C_MATCH, C_MATCH_EDGE = '#B3BCC2', '#8C989E'
Y_LO, Y_HI = 0.62, 520.0
DX, MS = 0.17, 3.4


def nd_colour(n):
    f = 0.30 + 0.60 * (np.asarray(n, float) - 1.0) / 9.0
    return CMAP(0.5 + 0.5 * f)


class _Ramp:
    pass


class _HandlerRamp(HandlerBase):
    """The legend key of the exact count, three heads in the colours it takes at
    one, four and ten driver nodes."""
    def create_artists(self, legend, orig, xd, yd, w, h, fontsize, trans):
        return [Line2D([xd + w * (0.18 + 0.32 * j)], [yd + h / 2], ls='none', marker='o', ms=MS,
                       mew=0.6, mec=NEUTRAL, mfc=nd_colour(n), transform=trans)
                for j, n in enumerate((1, 4, 10))]


def frame(ax, side):
    for s_ in ('left', 'right', 'top'):
        ax.spines[s_].set_visible(False)
    ax.spines['bottom'].set_linewidth(0.7)
    ax.grid(True, axis='y', alpha=0.40, linestyle=':', linewidth=0.5, color='gray')
    ax.set_axisbelow(True)
    if side == 'right':
        ax.yaxis.tick_right(); ax.yaxis.set_label_position('right')
    ax.tick_params(axis='y', which='both', left=(side == 'left'), right=(side == 'right'),
                   labelleft=(side == 'left'), labelright=(side == 'right'),
                   length=0, labelsize=6.0, pad=1.2)
    ax.tick_params(axis='x', which='major', length=2.4, width=0.6, direction='out',
                   labelsize=6.0, pad=1.2)
    ax.tick_params(axis='x', which='minor', length=1.3, width=0.5, direction='out')


axS = []
for i, kind in enumerate(KINDS):
    ax = box(LX[i], ST_Y, COL_W, ST_H)
    for arr, dx, exact in ((ens[kind]['match'], -DX, False), (ens[kind]['exact'], DX, True)):
        v, cnt = np.unique(arr, return_counts=True)
        ax.vlines(v + dx, Y_LO, cnt, color=NEUTRAL if exact else C_MATCH_EDGE, lw=0.8, zorder=2)
        ax.scatter(v + dx, cnt, s=MS ** 2, marker='o',
                   c=[nd_colour(x) for x in v] if exact else C_MATCH,
                   edgecolors=NEUTRAL if exact else C_MATCH_EDGE, linewidths=0.6, zorder=3,
                   clip_on=False)
    ax.set_yscale('log'); ax.set_ylim(Y_LO, Y_HI); ax.set_xlim(0.35, 10.65)
    ax.set_yticks([1, 10, 100]); ax.set_yticklabels(['1', '10', '100'])
    ax.yaxis.set_minor_locator(NullLocator())
    # three labelled ticks, the least, the middle and the largest count, as on
    # the count axis, and an unlabelled mark at every other value
    ax.set_xticks([1, 5, 10]); ax.set_xticklabels(['1', '5', '10'])
    ax.xaxis.set_minor_locator(FixedLocator([2, 3, 4, 6, 7, 8, 9]))
    frame(ax, 'right' if MIRROR[i] else 'left')
    e = ens[kind]
    ax.text(0.975, 0.955, f'$2^H$ = {effective_dof(e["H"].mean()):.1f}\n'
            f'$N_D$ = {e["match"].mean():.2f}, {e["exact"].mean():.2f}',
            transform=ax.transAxes, fontsize=6.0, ha='right', va='top', linespacing=1.35,
            zorder=9, bbox=dict(boxstyle='square,pad=0.10', fc='white', ec='none', alpha=0.95))
    ax.set_title(f'{N_REAL} realisations', fontsize=6.4, color=TXT, pad=3)
    ax.set_xlabel('driver nodes  $N_D$', fontsize=6.4, labelpad=1.5)
    axS.append(ax)
axS[0].set_ylabel('Frequency', fontsize=6.4, labelpad=1.5)
axS[3].set_ylabel('Frequency', fontsize=6.4, labelpad=2.0)
axS[0].legend(handles=[Line2D([], [], ls='none', marker='o', ms=MS, mew=0.6, mec=C_MATCH_EDGE,
                              mfc=C_MATCH), _Ramp()],
              labels=['matching', 'exact'], handler_map={_Ramp: _HandlerRamp()},
              loc='lower right', bbox_to_anchor=(1.0, 0.04), fontsize=6.0,
              handlelength=2.0, handleheight=0.8, handletextpad=0.4, borderaxespad=0.0,
              labelspacing=0.3, frameon=True, fancybox=False, framealpha=1.0,
              facecolor='white', edgecolor='none', borderpad=0.3)

for i, kind in enumerate(KINDS):
    heading('ABCD'[i], LX[i] + COL_W if MIRROR[i] else LX[i], MIRROR[i], CX[i], LABEL[kind],
            GEN[kind], T_Y)

# ---------------- F, influence against share ---------------------------------
# Every element of the structures drawn above, in the frame of the lower row of
# Figure 4, a line above and below and none at the sides. The scale-free
# elements are drawn as in D, coloured by share and sized by their links. Every
# lattice element sits on one point, drawn at the size of a lattice element. The
# small-world and random structures are drawn by their least-squares lines in the
# blue-grey tones of Figure 4, and the key, at the lower right where no element
# falls, gives the mean correlation over the 100 realisations of each ensemble.
# Each axis carries three labelled ticks, as the count axes of A to D do, and the
# axis at the right, facing G, gives the same values relative to the lattice.
SB_W = 1.850                               # width of each schematic, E and G
F_X, F_W = 2.650, 1.800
axf = box(F_X, BR_Y, F_W, BR_H)
axf.axvline(1.0 / N, color=INK, lw=0.7, ls=(0, (3.2, 2.2)), zorder=1)
axf.text(1.0 / N + 0.0006, 1.690, '$1/N$', fontsize=6.0, ha='left', va='top')
F_TONE = {'small-world': ENS_TONE['small-world'], 'random': ENS_TONE['random'], 'scalefree': INK}
for kind in ('small-world', 'random', 'scalefree'):
    d = INFL[kind]; xs_ = np.linspace(d['p'].min(), d['p'].max(), 50)
    axf.plot(xs_, np.polyval(d['fit'], xs_), color=F_TONE[kind], lw=1.0 if kind != 'scalefree' else 0.9,
             zorder=3, solid_capstyle='round')
sf = INFL['scalefree']; order_f = np.argsort(K_SF)
axf.scatter(sf['p'][order_f], sf['ac'][order_f], s=nsize(SHOW['scalefree'])[order_f],
            c=node_colour(sf['p'], N)[order_f], edgecolors=NEUTRAL, linewidths=0.3, zorder=4)
lat = INFL['regular']
axf.scatter([lat['p'][0]], [lat['ac'][0]], s=nsize(SHOW['regular'])[0], c=node_colour([lat['p'][0]], N),
            marker='o', edgecolors=INK, linewidths=0.8, zorder=6)
axf.text(lat['p'][0], lat['ac'][0] + 0.030, 'lattice', fontsize=6.0, ha='center', va='bottom', zorder=7,
         bbox=dict(boxstyle='square,pad=0.06', fc='white', ec='none', alpha=0.95))
axf.legend(handles=[Line2D([], [], color=F_TONE[k], lw=1.0) for k in ('small-world', 'random', 'scalefree')],
           labels=[f'{LABEL[k]}  {np.nanmean(ens[k]["r"]):.2f}' for k in ('small-world', 'random', 'scalefree')],
           title='mean $r$', title_fontsize=5.4, fontsize=5.4,
           loc='lower right', bbox_to_anchor=(1.0, 0.02), handlelength=1.3, handletextpad=0.4,
           labelspacing=0.2, borderaxespad=0.1, frameon=True, fancybox=False, framealpha=1.0,
           facecolor='white', edgecolor='none', borderpad=0.2, alignment='left')
axf.set_xlim(0, 0.052); axf.set_ylim(0.95, 1.70)
# three labelled ticks on each axis, the least, the middle and the largest value,
# as on the count axes of A to D, and an unlabelled mark at every other 0.005
axf.set_xticks([0, 0.025, 0.05]); axf.set_xticklabels(['0', '0.025', '0.05'])
axf.xaxis.set_minor_locator(FixedLocator([0.005, 0.010, 0.015, 0.020, 0.030, 0.035, 0.040, 0.045]))
axf.set_yticks([1.0, 1.3, 1.6]); axf.set_yticklabels(['1.0', '1.3', '1.6'])
axf.set_xlabel('coupling share  $p_i$', fontsize=6.4, labelpad=1.6)
axf.set_ylabel('average controllability', fontsize=6.4, labelpad=2.0)
# at the right, the same values relative to an element of the lattice, the evenly
# coupled structure, so that 1 marks even influence as 1/N marks even share. The
# axis has a short line of its own, since no dotted rule marks its values
AC_LAT = float(lat['ac'][0])
axf_r = axf.secondary_yaxis('right', functions=(lambda y: y / AC_LAT, lambda r: r * AC_LAT))
axf_r.set_yticks([0.8, 1.0, 1.2]); axf_r.set_yticklabels(['0.8', '1.0', '1.2'])
axf_r.set_ylabel('relative to the lattice', fontsize=6.4, labelpad=2.0)
axf_r.spines['right'].set_bounds(0.8, 1.2); axf_r.spines['right'].set_linewidth(0.6)
axf_r.tick_params(axis='y', length=2.4, width=0.6, direction='out', labelsize=6.0, pad=1.4)
for s_ in ('left', 'right'):
    axf.spines[s_].set_visible(False)
for s_ in ('top', 'bottom'):
    axf.spines[s_].set_visible(True); axf.spines[s_].set_linewidth(0.7)
axf.tick_params(axis='y', length=0, labelsize=6.0, pad=1.4)
axf.tick_params(axis='x', which='major', length=2.4, width=0.6, direction='out', labelsize=6.0, pad=1.4)
axf.tick_params(axis='x', which='minor', length=1.3, width=0.5, direction='out')
axf.grid(axis='y', alpha=0.40, linestyle=':', linewidth=0.5, color='gray', zorder=0)
axf.set_axisbelow(True)
heading('F', F_X - 0.340, False, F_X + F_W / 2, 'Influence against share', None, TB_Y)

# ---------------- E and G, one coordinative structure at two stages ----------
# Sixteen neural populations in the brain and eight muscles of the arm, drawn in
# the manner of A to D: each element coloured by its own eigenvector share on
# the scale of the structures above and sized by its links, a muscle by the
# width of its spindle, the links in grey, and the one driver node that full
# control needs ringed and fed by an input u. The muscles are spindles on the
# bones they move. The same twenty-four elements hold forty-eight links at both
# stages, spread evenly at the initial-like stage and laid down by preferential
# attachment at the skilled-like stage. The hand then reaches five times for a
# target T. Each path leaves the wrist along the hand, so it begins on the arm,
# passes through its endpoint x_i and runs on a little past it along the same
# curve, as a hand moves on after contact. At the initial-like stage the paths
# and their endpoints scatter about the target, and a double arrow marks the
# absolute error, AE = sum |x_i - T| / n, from the target to the farthest
# endpoint. At the skilled-like stage the paths still vary on the way but pass
# together through the centre of the target, so the error is too short to draw
# as an arrow and only its label is kept. Each label sits clear of the paths
# that run on. The paths are sketched rather than simulated. Drawn for
# illustration, not measured. Positions are in inches inside each box, with the
# figure facing right.
from scipy.interpolate import splprep, splev
from matplotlib.path import Path as MPath

BRAIN_FILL, BRAIN_EDGE, SULCUS = '#EEF1F3', '#C9D1D5', '#DCE1E4'
BODY, BONE, JOINT_EDGE, SCHEME_LINK = '#F0F2F4', '#D3DADE', '#9AA6AD', '#AAB4BA'
TARGET_FILL = '#F4F6F7'
AE_MIN = 0.040                                           # inches
FOLLOW = 0.100                                           # path beyond each endpoint, inches
SB_Y = 0.050
SB_H = TB_Y - 0.135 - SB_Y
BC, BRX, BRY = (0.800, SB_H - 0.330), 0.545, 0.300      # brain centre and half sizes
STEM = (BC[0] - 0.08 * BRX, BC[1] - 1.10 * BRY)
WAIST = (0.755, SB_H - 1.255)


def _toward(a, deg, length):
    return (a[0] + length * np.cos(np.radians(deg)), a[1] + length * np.sin(np.radians(deg)))


SHO, ELB = (0.805, SB_H - 0.755), (1.010, SB_H - 1.165)
WRI = _toward(ELB, 42.0, 0.300)                          # the forearm points at the target
HND = _toward(WRI, 44.0, 0.090)
TGT, TGT_R = _toward(HND, 46.0, 0.450), 0.065             # the target T


def _on(a, b, t, off, front):
    """A point a fraction t along the bone from a to b, off inches to its front
    or its back, the figure facing right."""
    a, b = np.array(a), np.array(b); u = (b - a) / np.hypot(*(b - a))
    n = np.array([-u[1], u[0]]) if front else np.array([u[1], -u[0]])
    return tuple(a + t * (b - a) + off * n)


# muscle centre, spindle length and the bone it lies along
MUSCLE_XY = {'MD': (_on(SHO, ELB, -0.09, 0.000, True), 0.085, (SHO, ELB)),
             'AD': (_on(SHO, ELB, 0.12, 0.088, True), 0.100, (SHO, ELB)),
             'PD': (_on(SHO, ELB, 0.12, 0.088, False), 0.100, (SHO, ELB)),
             'BL': (_on(SHO, ELB, 0.43, 0.078, True), 0.130, (SHO, ELB)),
             'BS': (_on(SHO, ELB, 0.74, 0.070, True), 0.115, (SHO, ELB)),
             'TR': (_on(SHO, ELB, 0.58, 0.075, False), 0.170, (SHO, ELB)),
             'BR': (_on(ELB, WRI, 0.36, 0.058, True), 0.150, (ELB, WRI)),
             'FL': (_on(ELB, WRI, 0.62, 0.058, False), 0.150, (ELB, WRI))}


def _angle(a, b):
    return np.degrees(np.arctan2(b[1] - a[1], b[0] - a[0]))


def _brain(ax):
    """A side view of the brain facing right: cerebrum, cerebellum and stem."""
    rel = np.array([(1.00, 0.05), (0.86, 0.55), (0.46, 0.93), (-0.08, 1.00), (-0.58, 0.86),
                    (-0.93, 0.42), (-1.00, -0.08), (-0.82, -0.44), (-0.42, -0.56), (0.04, -0.70),
                    (0.46, -0.62), (0.80, -0.36)])
    pts = np.vstack([rel, rel[:1]])
    tck, _ = splprep([pts[:, 0], pts[:, 1]], s=0, per=True)
    xx, yy = splev(np.linspace(0, 1, 240), tck)
    ax.add_patch(Ellipse((BC[0] - 0.50 * BRX, BC[1] - 0.66 * BRY), 0.92 * BRX, 0.60 * BRY,
                         facecolor=BRAIN_FILL, edgecolor=BRAIN_EDGE, lw=0.6, zorder=0))     # cerebellum
    ax.plot([BC[0] - 0.12 * BRX, STEM[0]], [BC[1] - 0.62 * BRY, STEM[1]], color=BRAIN_EDGE,
            lw=0.075 * 72, solid_capstyle='round', zorder=0)                                    # stem
    ax.fill(BC[0] + BRX * np.array(xx), BC[1] + BRY * np.array(yy), facecolor=BRAIN_FILL,
            edgecolor=BRAIN_EDGE, lw=0.6, zorder=1)                                             # cerebrum
    for curve in (((0.06, 0.99), (-0.04, 0.55), (-0.16, 0.12)),        # central sulcus
                  ((0.58, -0.24), (0.15, -0.12), (-0.38, -0.02)),      # lateral sulcus
                  ((0.70, 0.52), (0.42, 0.40), (0.20, 0.52))):         # a frontal sulcus
        c_ = np.array(curve); tck_, _ = splprep([c_[:, 0], c_[:, 1]], s=0, k=2)
        cx_, cy_ = splev(np.linspace(0, 1, 40), tck_)
        ax.plot(BC[0] + BRX * np.array(cx_), BC[1] + BRY * np.array(cy_), color=SULCUS, lw=0.6,
                solid_capstyle='round', zorder=1)


def reach(end, bend, lead=0.17, pull=0.45):
    """One hand path, a cubic curve that leaves the wrist along the hand, so that
    its first stretch lies on the arm, and arrives at the endpoint turned by bend
    degrees from the straight approach. lead sets how far the path keeps to the
    hand and pull how wide it swings before the endpoint."""
    p0 = np.array(WRI); d0 = np.array(HND) - p0; d0 = d0 / np.hypot(*d0)
    p1 = p0 + lead * d0
    p3 = np.array(end); v = p3 - p1; L = np.hypot(*v); v = v / L
    c, s = np.cos(np.radians(bend)), np.sin(np.radians(bend))
    p2 = p3 - pull * L * np.array([c * v[0] - s * v[1], s * v[0] + c * v[1]])
    return MPath([p0, p1, p2, p3], [MPath.MOVETO, MPath.CURVE4, MPath.CURVE4, MPath.CURVE4])


def _blossom(P, u1, u2, u3):
    """The polar form of the cubic with control points P, so that the same curve
    can be carried past its last control point without changing its shape."""
    Q = [(1 - u1) * P[i] + u1 * P[i + 1] for i in range(3)]
    R = [(1 - u2) * Q[i] + u2 * Q[i + 1] for i in range(2)]
    return (1 - u3) * R[0] + u3 * R[1]


def reach_through(end, bend, lead, pull, follow):
    """The hand path of reach, carried on past its endpoint along the same cubic
    for an arc length of follow inches, as a hand moves on a little after contact."""
    P = [np.array(v) for v in reach(end, bend, lead, pull).vertices]
    def ext(d):
        e = 1 + d
        return [P[3], _blossom(P, 1, 1, e), _blossom(P, 1, e, e), _blossom(P, e, e, e)]
    def length(d):
        C = ext(d); t = np.linspace(0, 1, 60)[:, None]
        pts = (1 - t) ** 3 * C[0] + 3 * (1 - t) ** 2 * t * C[1] + 3 * (1 - t) * t ** 2 * C[2] + t ** 3 * C[3]
        return np.hypot(*np.diff(pts, axis=0).T).sum()
    lo, hi = 0.0, 1.0
    for _ in range(50):                                   # the extension of the requested length
        mid = 0.5 * (lo + hi)
        lo, hi = (mid, hi) if length(mid) < follow else (lo, mid)
    C = ext(0.5 * (lo + hi))
    return MPath([P[0], P[1], P[2], P[3], C[1], C[2], C[3]],
                 [MPath.MOVETO] + [MPath.CURVE4] * 6)


NEURAL_XY = {n_: (BC[0] + BRX * v[0], BC[1] + BRY * v[1]) for n_, v in NEURAL_REL.items()}
_KMAX_S = float(max(max(dict(SCHEME[k]['G'].degree()).values()) for k in SCHEME))


def draw_scheme(x0, name, ends, bends, formula, mirror=False, leads=None, pulls=None):
    S = SCHEME[name]; G = S['G']
    ax = box(x0, SB_Y, SB_W, SB_H)
    # the skilled-like stage is drawn as the mirror image of the initial-like
    # stage, the figure facing left, by running its horizontal axis from right to left
    ax.set_xlim((SB_W, 0) if mirror else (0, SB_W)); ax.set_ylim(0, SB_H)
    ax.set_aspect('equal'); ax.axis('off')
    # body, brain and arm
    ax.plot([STEM[0], WAIST[0]], [STEM[1] - 0.02, WAIST[1]], color=BODY, lw=0.120 * 72,
            solid_capstyle='round', zorder=0)
    _brain(ax)
    kw = dict(color=BONE, solid_capstyle='round', solid_joinstyle='round', zorder=1)
    ax.plot(*zip(SHO, ELB), lw=0.040 * 72, **kw)
    ax.plot(*zip(ELB, WRI, HND), lw=0.034 * 72, **kw)
    # the target, the reaches and their endpoints
    ax.add_patch(plt.Circle(TGT, TGT_R, facecolor=TARGET_FILL, edgecolor=NEUTRAL, lw=0.6, zorder=2))
    ax.plot([TGT[0]], [TGT[1]], marker='+', ms=3.4, mew=0.6, color=NEUTRAL, zorder=2.1)
    xs_ = np.array([(TGT[0] + dx_, TGT[1] + dy_) for dx_, dy_ in ends])
    leads = leads or (0.17,) * len(bends); pulls = pulls or (0.45,) * len(bends)
    heading_at = []                                       # the direction of each path as it passes its endpoint
    for e_, b_, l_, q_ in zip(xs_, bends, leads, pulls):
        ax.add_patch(FancyArrowPatch(path=reach_through(e_, b_, l_, q_, FOLLOW), arrowstyle='-|>',
                                     mutation_scale=3.6, color=NEUTRAL, lw=0.55, alpha=0.75, shrinkA=0,
                                     shrinkB=1.1, zorder=2.5))
        P_ = reach(e_, b_, l_, q_).vertices; t_ = P_[3] - P_[2]; heading_at.append(t_ / np.hypot(*t_))
    ax.scatter(xs_[:, 0], xs_[:, 1], s=4.0, c=INK, linewidths=0, zorder=2.6)
    for jx in (SHO, ELB, WRI):
        ax.scatter([jx[0]], [jx[1]], s=9, c='white', edgecolors=JOINT_EDGE, linewidths=0.5, zorder=2.7)
    # the absolute error, from the target to the farthest endpoint
    far = xs_[np.argmax(np.hypot(*(xs_ - np.array(TGT)).T))]
    if np.hypot(*(far - np.array(TGT))) > AE_MIN:          # too short to draw as a double arrow
        ax.add_patch(FancyArrowPatch(TGT, tuple(far), arrowstyle='<|-|>', mutation_scale=3.0, color=INK,
                                     lw=0.5, shrinkA=0, shrinkB=0, zorder=2.8))
    v_ = far - np.array(TGT); v_ = v_ / np.hypot(*v_); n_ = np.array([-v_[1], v_[0]])
    if np.dot(n_, np.array(TGT) - np.array(HND)) < 0:
        n_ = -n_                                          # the label on the side away from the hand
    span = np.hypot(*(far - np.array(TGT)))
    if span > TGT_R + 0.03:
        # beside the farthest endpoint, turned away from the path that runs on through it
        h_ = heading_at[int(np.argmax(np.hypot(*(xs_ - np.array(TGT)).T)))]
        turns = [np.array([np.cos(a_) * v_[0] - np.sin(a_) * v_[1], np.sin(a_) * v_[0] + np.cos(a_) * v_[1]])
                 for a_ in np.radians([50.0, -50.0])]
        lab = far + 0.080 * min(turns, key=lambda c_: float(np.dot(c_, h_)))
    else:
        # above the target and toward the hand, between the paths that arrive and those that run on
        lab = np.array(TGT) + np.array([-0.112, 0.094])
    ax.text(*lab, '$AE$', fontsize=6.0, ha='center', va='center', zorder=8)
    if formula:
        ax.text(SB_W, 0.010, r'$AE = \Sigma\,|x_i - T|\,/\,n$', fontsize=6.0,
                ha='left' if mirror else 'right', va='bottom', zorder=8)
    # the structure: links, muscles, populations
    nodes_ = list(G.nodes()); k_ = dict(G.degree())
    xy_ = {**NEURAL_XY, **{m_: MUSCLE_XY[m_][0] for m_ in MUSCLE}}
    ax.add_collection(LineCollection([(xy_[a_], xy_[b_]) for a_, b_ in G.edges()], colors=SCHEME_LINK,
                                     linewidths=0.38, alpha=0.85, zorder=3))
    col_ = dict(zip(nodes_, node_colour(S['p'], len(nodes_))))
    for m_ in MUSCLE:
        (cx_, cy_), ln, bone = MUSCLE_XY[m_]
        ax.add_patch(Ellipse((cx_, cy_), 0.032 + 0.0045 * k_[m_], ln, angle=_angle(*bone) - 90,
                             facecolor=col_[m_], edgecolor=NEUTRAL, lw=0.35, zorder=4))
    size_ = {n_: 5.0 + 40.0 * k_[n_] / _KMAX_S for n_ in NEURAL}
    xyn = np.array([NEURAL_XY[n_] for n_ in NEURAL])
    ax.scatter(xyn[:, 0], xyn[:, 1], s=[size_[n_] for n_ in NEURAL], c=[col_[n_] for n_ in NEURAL],
               edgecolors=NEUTRAL, linewidths=0.3, zorder=4)
    # the one driver node, ringed and fed by an input of its own, as in A to D
    d = S['driver']; p0 = np.array(NEURAL_XY[d])
    ax.scatter([p0[0]], [p0[1]], s=size_[d] + 30, facecolors='none', edgecolors=INK, linewidths=0.85,
               zorder=5)
    u = p0 - np.array(BC); u = u / np.hypot(*u)
    tip, z_end = p0 + u * R_TIP, p0 + u * R_ZIG
    z_start, q = p0 + u * (R_NODE - 0.030), p0 + u * R_NODE
    zz = zigzag(z_start, z_end, n=3, amp=0.0135)
    ax.plot(zz[:, 0], zz[:, 1], color=INK, lw=0.65, solid_joinstyle='miter', zorder=6, clip_on=False)
    ax.add_patch(FancyArrowPatch(tuple(z_end), tuple(tip), arrowstyle='-|>', mutation_scale=5.0,
                                 color=INK, lw=0.65, shrinkA=0, shrinkB=0, zorder=6, clip_on=False))
    ax.scatter([q[0]], [q[1]], s=11.0, c=INK, edgecolors=INK, linewidths=0.3, zorder=7, clip_on=False)
    ax.text(*(p0 + u * R_LAB), '$u$', fontsize=6.4, ha='center', va='center', zorder=7, clip_on=False)
    ax.text(-0.050, 0.150, f'$2^H$ = {S["eff"]:.1f} of {len(nodes_)}\n$N_D$ = {S["nd"]}', fontsize=6.0,
            ha='right' if mirror else 'left', va='center', linespacing=1.45,
            multialignment='right' if mirror else 'left')
    return ax


# reaches: where each ends, relative to the target, how far it turns on the way
# and, at the skilled-like stage, how long it keeps to the hand and how wide it swings
AX_INI = draw_scheme(0.250, 'initial',
                     ends=((-0.110, -0.077), (-0.066, 0.121), (0.132, 0.110), (0.143, -0.022),
                           (0.044, -0.132)),
                     bends=(-40.0, 35.0, 15.0, -10.0, -35.0), formula=True)
AX_SKL = draw_scheme(FIG_W_IN - 0.250 - SB_W, 'skilled',
                     ends=((-0.003, 0.001), (0.001, -0.003), (0.002, 0.002), (-0.002, -0.002),
                           (0.000, 0.003)),
                     bends=(-40.0, -26.0, -6.0, 17.0, 34.0), formula=False, mirror=True,
                     leads=(0.16, 0.18, 0.15, 0.19, 0.17), pulls=(0.80, 0.60, 0.62, 0.55, 0.76))
heading('E', LX[0], False, 0.250 + SB_W / 2, 'Initial-like', 'coupling spread evenly', TB_Y)
heading('G', LX[3] + COL_W, True, FIG_W_IN - 0.250 - SB_W / 2, 'Skilled-like',
        'grown by preferential attachment', TB_Y)

plt.show()
```
